# AI4I 2020 Predictive Maintenance — Data Exploration

This notebook explores the AI4I 2020 Predictive Maintenance Dataset before preprocessing and model training.

The goal is to understand the dataset structure, examine the available features, check for missing values, and investigate the distribution of machine failures.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../Data/ai4i2020.csv")
```

```python
df.head()
```

## Dataset Structure

The dataset has been successfully loaded. The next step is to examine its size, column names, data types, and whether any values are missing.

```python
print("Dataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())
```

## Target Distribution

The target variable for this project is `Machine failure`. Examining the number of failures and non-failures helps determine whether the dataset is balanced.

```python
print("Machine Failure Counts:")
print(df["Machine failure"].value_counts())

print("\nMachine Failure Percentages:")
print(df["Machine failure"].value_counts(normalize=True) * 100)
```

## Product Type Distribution

The `Type` feature represents the product quality category: Low, Medium, or High. This section examines how often each product type appears in the dataset.

```python
print("Product Type Counts:")
print(df["Type"].value_counts())

print("\nProduct Type Percentages:")
print(df["Type"].value_counts(normalize=True) * 100)
```

```python
df["Machine failure"].value_counts().plot(kind="bar")

plt.title("Machine Failure Distribution")
plt.xlabel("Machine Failure")
plt.ylabel("Count")
plt.xticks([0, 1], ["No Failure", "Failure"], rotation=0)
plt.show()
```
